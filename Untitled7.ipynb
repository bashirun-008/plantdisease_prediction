```python
import os

# Let's find the exact path to the images
base_path = "plant_data"
subfolders = os.listdir(base_path)

# If there's a nested folder like 'PlantVillage', we find it dynamically
if len(subfolders) == 1:
    DATA_DIR = os.path.join(base_path, subfolders[0])
else:
    DATA_DIR = base_path

# Print out some of the plant disease classes
print("Found data directory:", DATA_DIR)
class_names = sorted(os.listdir(DATA_DIR))
print(f"\nTotal classes found: {len(class_names)}")
print("First 5 classes look like this:", class_names[:5])
```

```python
import tensorflow as tf

# Explicitly set the path to the main folder containing the actual plant folders
DATA_DIR = "plant_data/PlantVillage"
IMAGE_SIZE = (224, 224) # Standard input size for modern image models
BATCH_SIZE = 32

# Load and split into Training Dataset
train_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="training",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

# Load and split into Validation Dataset
val_ds = tf.keras.utils.image_dataset_from_directory(
    DATA_DIR,
    validation_split=0.2,
    subset="validation",
    seed=123,
    image_size=IMAGE_SIZE,
    batch_size=BATCH_SIZE
)

# Grab the actual class names (e.g., 'Tomato___Early_blight')
class_names = train_ds.class_names
print(f"\nSuccessfully loaded {len(class_names)} unique plant/disease classes!")
```

```python
from tensorflow.keras import layers

# Create data augmentation layers
data_augmentation = tf.keras.Sequential([
    layers.RandomFlip("horizontal_and_vertical"),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.1),
])

# Optimize data loading performance (pre-fetches images in the background)
AUTOTUNE = tf.data.AUTOTUNE
train_ds = train_ds.map(lambda x, y: (data_augmentation(x, training=True), y)).prefetch(buffer_size=AUTOTUNE)
val_ds = val_ds.prefetch(buffer_size=AUTOTUNE)

print("Data augmentation and performance optimization complete!")
```

```python
from tensorflow.keras import models

# 1. Load the pre-trained MobileNetV2 without its final classification layer
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(224, 224, 3),
    include_top=False,
    weights='imagenet'
)

# 2. Freeze the pre-trained weights so we don't destroy Google's learned features
base_model.trainable = False

# 3. Assemble our final model architecture
model = models.Sequential([
    # MobileNetV2 expects pixel values between -1 and 1
    layers.Rescaling(1./127.5, offset=-1),

    # The pre-trained core
    base_model,

    # Flatten the features into a single vector
    layers.GlobalAveragePooling2D(),

    # A little dropout to prevent overfitting
    layers.Dropout(0.2),

    # The final output layer: 15 nodes for our 15 plant disease classes
    layers.Dense(len(class_names), activation='softmax')
])

# 4. Compile the model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

print("Model successfully built and compiled!")
model.summary()  # This will print the internal structure of your model
```

```python
# Start the training process
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=5
)
```

```python
import numpy as np
import matplotlib.pyplot as plt

def predict_plant_disease(image_path):
    # 1. Load image and resize to target 224x224
    img = tf.keras.utils.load_img(image_path, target_size=IMAGE_SIZE)

    # 2. Convert to array and add batch dimension (1, 224, 224, 3)
    img_array = tf.keras.utils.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)

    # 3. Predict class probabilities
    predictions = model.predict(img_array)

    # 4. Extract highest probability class name and confidence
    predicted_class_idx = np.argmax(predictions[0])
    predicted_class_name = class_names[predicted_class_idx]
    confidence = 100 * np.max(predictions[0])

    # 5. Plot the leaf alongside prediction results
    plt.figure(figsize=(5, 5))
    plt.imshow(img)
    plt.axis("off")
    plt.title(f"Prediction: {predicted_class_name}\nConfidence: {confidence:.2f}%")
    plt.show()

    print(f"🌱 Result: **{predicted_class_name}**")
    print(f"🎯 Confidence: {confidence:.2f}%")
```

```python
!pip install gradio -q
import gradio as gr
import numpy as np
import tensorflow as tf

# 1. The main prediction logic
def predict_leaf(image):
    if image is None:
        return None

    # Resize image to match model input shape (224x224)
    img_resized = tf.image.resize(image, (224, 224))
    img_array = tf.expand_dims(img_resized, axis=0)

    # Predict probabilities for all classes
    predictions = model.predict(img_array)[0]

    # Format as a dictionary for Gradio's Label output
    return {class_names[i]: float(predictions[i]) for i in range(len(class_names))}

# 2. Build a clean UI using Gradio's Soft Theme
theme = gr.themes.Soft(
    primary_hue="emerald",  # Clean botanical green tone
    radius_size="md"        # Soft rounded corners
)

# 3. Create the Interface (flagging_mode="never" used for modern Gradio)
demo = gr.Interface(
    fn=predict_leaf,
    inputs=gr.Image(type="numpy", label="Upload Leaf Image"),
    outputs=gr.Label(num_top_classes=3, label="Top Predictions"),
    title="🌱 Plant Disease Detector",
    description="Upload a photo of a leaf to check for diseases using Transfer Learning.",
    theme=theme,
    flagging_mode="never"  # Updated argument name for newer Gradio versions
)

# 4. Launch the app
demo.launch(share=True)
```
